```python
import numpy as np

X = [0.25, 0.30, 0.35, 0.40, 0.45, 0.50, 0.55]
Y = [
    1.2557,
    2.1764,
    3.1218,
    4.0482,
    5.9875,
    6.9195,
    7.8359,
]

n = len(Y)
# Создаем таблицу разностей
F = np.zeros((n, n))
F[:, 0] = Y

# Заполняем разности
for j in range(1, n):
    for i in range(n - j):
        F[i, j] = F[i + 1, j - 1] - F[i, j - 1]

# Выводим таблицу с точностью 4 знака
print("\nТаблица конечных разностей:")
print("-" * 50)
for i in range(n):
    for j in range(n - i):
        print(f"{F[i, j]:10.4f}", end="")
    print()
```

```python
def lagrange_formula(x_target, x_points, y_points):
    """
    Вычисляет интерполяционный многочлен Лагранжа в точке x_target
    и выводит формулу в виде суммы.
    """
    n = len(x_points)
    terms = []
    coefficients = []

    print(f"\n{'='*60}")
    print(f"Интерполяция Лагранжа для x = {x_target}")
    print(f"{'='*60}\n")

    for i in range(n):
        # Вычисляем L_i(x)
        L = 1.0
        numerator_factors = []
        denominator_factors = []

        for j in range(n):
            if j != i:
                num = x_target - x_points[j]
                den = x_points[i] - x_points[j]
                L *= num / den

                numerator_factors.append(f"({x_target} - {x_points[j]})")
                denominator_factors.append(f"({x_points[i]} - {x_points[j]})")

        coefficients.append(L)

        # Формируем строку для текущего члена
        num_str = "·".join(numerator_factors)
        den_str = "·".join(denominator_factors)

        terms.append(f"{y_points[i]}·[{num_str}/{den_str}]")

    # Вывод формулы
    print("Формула P(x) =")
    formula = " + \n".join(terms)
    print(formula)
    print("\n" + "=" * 60)

    # Вывод коэффициентов
    print("\nКоэффициенты L_i(x) =")
    for i, (xi, Li) in enumerate(zip(x_points, coefficients)):
        print(f"L_{i}({x_target}) = {Li:.10f}  (узел x_{i} = {xi})")

    # Вычисление значения
    result = sum(y_points[i] * coefficients[i] for i in range(n))

    print(f"\nСумма: P({x_target}) = ", end="")
    print(" + ".join([f"{y_points[i]}·{coefficients[i]:.6f}" for i in range(n)]))
    print(f"\n{'='*60}")
    print(f"РЕЗУЛЬТАТ: P({x_target}) = {result:.8f}")
    print(f"{'='*60}")

    return result


# Данные из таблицы
x_nodes = [0.25, 0.30, 0.35, 0.40, 0.45, 0.50, 0.55]
y_nodes = [1.2557, 2.1764, 3.1218, 4.0482, 5.9875, 6.9195, 7.8359]

# Точки для интерполяции
X1 = 0.512
X2 = 0.372

# Вычисления
P_X1 = lagrange_formula(X1, x_nodes, y_nodes)
P_X2 = lagrange_formula(X2, x_nodes, y_nodes)

# Итог
print("\n\n" + "=" * 60)
print("ИТОГОВЫЕ ЗНАЧЕНИЯ")
print("=" * 60)
print(f"P({X1}) = {P_X1:.8f}")
print(f"P({X2}) = {P_X2:.8f}")
print("=" * 60)
```
