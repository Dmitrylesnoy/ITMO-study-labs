y=12x/(x^4+6)      ,     x∈[0,2]   h=0,2


```python
def func(x):
    return 12*x / (x**4 + 6)
```

```python
X=[]
Y=[]
for x in [(j*0.2) for j in range(int((2-0)/0.2 +1))] :
    print(f"X: {x}, Y: {func(x)}")
    X+=[x]
    Y+=[func(x)]
X, Y
```

```python
SX=sum(X)
SXX=sum(x**2 for x in X)
SY=sum(Y)
SXY=sum(X[i]*Y[i] for i in range(len(X)))

SX, SXX, SY, SXY
```

```python
d=SXX*11-SX*SX
d1=SXY*11-SX*SY
d2=SXX*SY-SX*SXY


d, d1, d2
```

```python
a=d1/d
b=d2/d


def P(x):
    return a * x + b


a, b
```

```python
for i in range(len(X)):
    x,y=X[i], Y[i]
    p=P(x)
    e=p-y
    print(f"i:{i}     X: {x:.3f}  Y:{y:.3f}   P:{p:.3f}   E:{e:.3f}")
```
